```python
import pandas as pd
```

```python
df_parls = pd.read_csv('../../data/df_parl_detalhes.csv')
df_ird = pd.read_csv('../../outputs/df_equality_index.csv')
df_for_ird = pd.read_csv('../../outputs/df_for_ird.csv')
df_parls_per_legislatura = pd.read_csv('../../data/df_parlamentares_por_legislatura.csv')
```

```python
df_parls_per_legislatura.loc[0]
```

```python
df = df_ird.merge(df_parls_per_legislatura[['id', 'idLegislatura', 'nome', 'siglaPartido', 'siglaUf']], left_on=['legislatura', 'parl_id'], right_on=['idLegislatura', 'id'], how='left')
del df['id']
del df['idLegislatura']

df = df.merge(df_for_ird[['id', 'idLegislatura', 'F', 'M', 'branca', 'n_branca', 'parda', 'preta', 'amarela', 'indigena']], left_on=['legislatura', 'parl_id'], right_on=['idLegislatura', 'id'])
del df['id']
del df['idLegislatura']
```

```python
df.to_csv('../../outputs/df_ird.csv', index=False)
```
