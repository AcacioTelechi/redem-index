```python
import pandas as pd
```

```python
COLS_TO_KEEP = [
    "pub_id",
    "date",
    "user_id",
    "screen_name",
    "text",
    "likes_count",
    "comment_count",
    'favorite_count',
    'retweet_count',
    'quote_count'
    'reply_count'
    "analisado",
    "is_fn",
]
```

```python
df_res = pd.read_excel("df_res_hs.xlsx")
df_pubs = pd.read_excel("./2023_Completo_redem_0304.xlsx")
```

```python
df_train = df_pubs.merge(
    df_res[["pub_id", "resumo_2", "analisado", "is_fn"]],
    left_on="Message-ID",
    right_on="pub_id",
    how="left",
)

df_train.rename(
    {
        "Profile-ID": "user_id",
        "Profile": "screen_name",
        "Date": "date",
        "Message": "text",
        "Number of comments": "comment_count",
        "Number of Reactions": "reactions_count",
        "Comments & Shares": "comments_and_reply_count",
        "Number of Likes": "likes_count",
    },
    axis=1,
    inplace=True
)
```

```python
df_covid = pd.read_excel("./Parlamentares e o coronavírus-Base Aos fatos.xlsx")
df_covid["is_fn"] = df_covid["avaliação"].apply(lambda x: 1 if x == "desinfo" else 0)
df_covid['analisado'] = 1
df_covid.rename(
    {
        "Unnamed: 0": "pub_id",
        "data": "date",
    },
    axis=1,
    inplace=True,
)
```

```python
pd.concat([df_train, df_covid])[COLS_TO_KEEP].to_excel('./data/df_train.xlsx')
```
