```python
from embeders import OllamaEmbeddingFunction

embedder = OllamaEmbeddingFunction()
```

```python
import numpy as np


a = embedder('O céu é azul')
not_a = embedder('O céu não é azul')
diff = np.array(a) - np.array(not_a)

b = embedder('O céu é vermelho')
new_b = np.array(b) + diff


print(a)
print(not_a)
print(a - not_a)
print(b)
print(new_b)

# from embed vector to text
embedder.client.generate(model=embedder.model, prompt=new_b)
```
