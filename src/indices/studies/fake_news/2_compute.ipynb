```python
import pandas as pd
```

```python
df_parls_raw = pd.read_csv("../../../../data/df_parlamentares_por_legislatura.csv")
```

```python
df_parls = df_parls_raw[(df_parls_raw['idLegislatura'] == 56) & (df_parls_raw['nome'].notna())]

df_parls = df_parls[['id', 'nome']].drop_duplicates()
```

```python

def is_in_divulgadores(name: str, divulgadores: str):
    return str(name).lower() in divulgadores.lower()


def count_in_divulgadores(name: str, divulgadores: str):
    return sum(1 for d in divulgadores if is_in_divulgadores(name, d))

df = pd.read_csv("./divulgadores_3.csv")

final = []
for i, parl in enumerate(df_parls['nome']):
    final.append({
        "nome": parl,
        "count": count_in_divulgadores(parl, df['divulgadores'])
    })

df_final = pd.DataFrame(final)
df_fn = df_final[df_final["count"] > 0]

print(
    f"Total de notícias analisadas: {len(df)}\n"
    f"Total de parlamentares: {len(df_final)}\nTotal de divulgadores: {len(df_fn)}\n% de divulgadores: {len(df_fn)/len(df_final)*100:.2f}%"
)

df_fn.sort_values(by="count", ascending=False)
```
